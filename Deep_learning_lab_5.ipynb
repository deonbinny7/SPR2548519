```python
data = open("/content/text.txt").read()

print(data[:500])
#Basically iam doing is printing the first 500 words in our dataset
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer()

tokenizer.fit_on_texts([data])

total_words = len(tokenizer.word_index) + 1

print("Total Words:", total_words)
#The computer cannot understand words, so we convert:These sentence into nnumbers,This is called Tokenization.
```

```python
input_sequences = []

for line in data.split('\n'):

    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

print(input_sequences[:10])
#What this code does
#data.split('\n')

# Splits the dataset line by line

# texts_to_sequences

# Converts words → numbers

# token_list[:i+1]

# Creates sequences
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

max_sequence_len = max([len(x) for x in input_sequences])

input_sequences = pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')

print(input_sequences[:10])

#We are actually doing padding here


# What this code does
# max_sequence_len
# Finds the longest sequence.

# pad_sequences
# Adds 0 in the beginning.
```

```python
xs = input_sequences[:,:-1] #Input words
ys = input_sequences[:,-1] #next word to predict
# Split Input (X) and Output (y)
```

```python
from tensorflow.keras.utils import to_categorical

ys = to_categorical(ys, num_classes=total_words)

print(ys.shape)

#From the output - Meaning that 601 training samples and 170 vocabulary samples
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model = Sequential()

model.add(Embedding(total_words, 100, input_length=max_sequence_len-1))
model.add(SimpleRNN(150))

model.add(Dense(total_words, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

model.summary()

#As in the output:
# Param  : 0
# (unbuilt)
# Total params: 0

# This means:

# The neural network architecture is defined, but
# TensorFlow has not initialized the weights yet.

# So the model is not built yet, which is why parameters show 0.
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model = Sequential()

model.add(Embedding(total_words, 100, input_length=max_sequence_len-1))
model.add(SimpleRNN(150))

model.add(Dense(total_words, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

model.build(input_shape=(None, max_sequence_len-1))
model.summary()

#Frome the output
# What each line means
# Embedding layer
#Converts words into 100-dimensional vectors.
#150 neurons that learn sequence patterns.
#Dense Layer
#Outputs probability for each word in vocabulary.
```

```python
from tensorflow.keras.layers import LSTM

model_lstm = Sequential()

model_lstm.add(Embedding(total_words, 100))
model_lstm.add(LSTM(150))
model_lstm.add(Dense(total_words, activation='softmax'))

model_lstm.compile(loss='categorical_crossentropy',
                   optimizer='adam',
                   metrics=['accuracy'])

model_lstm.build(input_shape=(None, max_sequence_len-1))

model_lstm.summary()

#LSTM Implementation
```

```python
model_lstm.fit(xs, ys, epochs=50)
#Training the lstm model
```

```python
model.fit(xs, ys, epochs=50, verbose=1)
```

```python
# What your training output shows

# Epoch 1: accuracy ≈ 0.03, loss ≈ 4.97

# Epoch 50: accuracy ≈ 0.71, loss ≈ 0.64

# Loss decreased → the model is making fewer mistakes

# Accuracy increased → the model is learning next-word patterns
```

```python
def generate_text(seed_text, next_words):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')

        predicted = model.predict(token_list, verbose=0)
        predicted_word_index = np.argmax(predicted)

        output_word = tokenizer.index_word[predicted_word_index]

        seed_text = seed_text + " " + output_word

    return seed_text
```

```python
print(generate_text("machine learning", 5))
```
